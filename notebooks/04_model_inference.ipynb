### Verify that the model works independently of the training notebook

```python
import joblib
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

MODEL_PATH = PROJECT_ROOT / "models" / "best_model.joblib"

model = joblib.load(MODEL_PATH)
```

```python
fake_student = pd.DataFrame([{
    # 20,Female,5,5.6,27.2,2.81,5.2,11.3,Moderate,5.4,8.7,87.8,No,Poor,Yes
    "Age": 20,
    "Gender": "Female",
    "Year_of_Study": 5,
    "Sleep_Hours": 5.6,
    "Study_Hours_Per_Week": None,
    "GPA": 2.81,
    "Social_Media_Hours": 5.2,
    "Screen_Time_Hours": 11.3,
    "Physical_Activity": "Moderate",
    "Financial_Stress": 5.4,
    "Family_Support": 8.7,
    "Attendance_Rate": 87.8,
    "Part_Time_Job": "No",
    "Diet_Quality": "Poor",
    "Previous_Mental_Health_History": "Yes"
}])

prediction = model.predict(fake_student)

print("Predicted risk : ", prediction[0])

probabiblities = model.predict_proba(fake_student)[0]

results = pd.DataFrame(
    {
        "Risk": model.classes_,
        "Probability": probabiblities
    }
)
display(results)
```
